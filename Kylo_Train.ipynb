# Kylo AI — آموزش آزمایشی از صفر 🖤

این نوت‌بوک مخزن Kylo AI را می‌گیرد، وابستگی‌ها را نصب می‌کند و آموزش را اجرا می‌کند. باید خودت سلول‌ها را اجرا کنی؛ من نمی‌توانم جلسهٔ Colab را از طرف تو روشن نگه دارم.

**مهم:** دیتاست پیش‌فرض کوچک است و فقط برای تست خط لولهٔ آموزش مناسب است؛ از آن انتظار یک چت‌بات قوی نداشته باش. برای نتیجهٔ بهتر، متن فارسی/انگلیسی بزرگ‌تر و مجاز به استفاده آماده کن. اجرای یک‌ساعته هم به‌تنهایی کیفیت را تضمین نمی‌کند.

```python
# 1) دریافت کد و نصب کتابخانه‌ها
import os, subprocess, sys
repo = '/content/Kylo-AI'
if not os.path.isdir(repo):
    subprocess.run(['git', 'clone', 'https://github.com/email1fje-dev/Kylo-AI.git', repo], check=True)
os.chdir(repo)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
print('آماده‌ایم. اگر Colab گزینهٔ GPU دارد، از Runtime > Change runtime type آن را انتخاب کن؛ دسترسی GPU تضمین‌شده نیست.')
```

## 2) (اختیاری) دیتاست خودت را بارگذاری کن
اگر فایل متنی بزرگ‌تر و مجاز برای آموزش داری، سلول بعد را اجرا کن و فایل `.txt` را انتخاب کن. اگر این مرحله را رد کنی، از دیتاست نمونه استفاده می‌شود.

```python
from google.colab import files
uploaded = files.upload()  # یک فایل متنی .txt انتخاب کن؛ برای رد کردن، سلول را اجرا نکن
for name, data in uploaded.items():
    if name.lower().endswith('.txt'):
        with open('data/train.txt', 'wb') as f:
            f.write(data)
        print('دیتاست انتخاب‌شده:', name)
        break
else:
    print('فایل txt انتخاب نشد؛ دیتاست نمونه استفاده می‌شود.')
```

## 3) اجرای آموزش ۶۰ دقیقه‌ای
این سلول ممکن است طولانی باشد. اتصال Colab می‌تواند قطع شود و سهمیهٔ رایگان محدود است. مدل فعلی یک نمونهٔ کاراکتری آموزشی است.

```python
data_path = 'data/train.txt' if os.path.isfile('data/train.txt') else 'data/starter_corpus.txt'
print('Dataset:', data_path)
subprocess.run([sys.executable, 'train.py', '--data', data_path, '--minutes', '60', '--save-every', '100'], check=True)
```

## 4) دانلود checkpoint
اگر آموزش با موفقیت تمام شد، فایل مدل را دانلود کن. برای استفاده در سرور، باید آن را به محیط اجرا منتقل کنی؛ آن را در مخزن عمومی GitHub قرار نده.

```python
from google.colab import files
checkpoint = 'checkpoints/kylo.pt'
if os.path.isfile(checkpoint):
    files.download(checkpoint)
else:
    print('Checkpoint پیدا نشد؛ اول سلول آموزش را با موفقیت اجرا کن.')
```
